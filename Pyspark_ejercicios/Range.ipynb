```python
import findspark
findspark.init()
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
```

```python
from pyspark.sql.types import IntegerType, LongType
```

```python
numbers  = [(1,), (2,), (4,), (10,)]
df_number = spark.createDataFrame(numbers,["Number"])
# df_number = spark.sparkContext.parallelize(numbers).toDF()
```

```python
full_range = spark.range(1, 11).toDF("Number")
```

```python
full_range.show()
```

con left_anti, hace el join con las numeros que no tiene el df_numbers, así obtenemos los datos que no están en un df

Recuerda que el left_anti join es útil para identificar registros que existen en un DataFrame pero no en otro. Es una herramienta poderosa para depurar datos y encontrar discrepancias entre conjuntos de datos.

```python
missin_numbers = full_range.join(df_number, "Number", "left_anti")
missin_numbers.show()
```
